```python
# ============================================================
# VISUAL TRAFFIC SIGNAL SIMULATION
# Q-LEARNING
# ============================================================

import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np
import random

# Initialize Q-table (States x Actions)
# States: 0=LOW TRAFFIC, 1=MEDIUM TRAFFIC, 2=HIGH TRAFFIC
# Actions: 0=KEEP SIGNAL, 1=CHANGE SIGNAL
q_table = np.zeros((3, 2))

# ------------------------------------------------------------
# Create Traffic Intersection
# ------------------------------------------------------------

fig, ax = plt.subplots(figsize=(12, 8))

ax.set_xlim(0, 100)
ax.set_ylim(0, 100)
ax.set_aspect("equal")

# Background
ax.set_facecolor("lightgreen")

# ------------------------------------------------------------
# Roads
# ------------------------------------------------------------

# Horizontal road
horizontal_road = patches.Rectangle(
    (0, 35), 100, 30,
    facecolor="dimgray"
)

# Vertical road
vertical_road = patches.Rectangle(
    (35, 0), 30, 100,
    facecolor="dimgray"
)

ax.add_patch(horizontal_road)
ax.add_patch(vertical_road)

# ------------------------------------------------------------
# Road Lane Lines
# ------------------------------------------------------------

# Horizontal center line
ax.plot(
    [0, 100],
    [50, 50],
    linestyle="--",
    linewidth=2
)

# Vertical center line
ax.plot(
    [50, 50],
    [0, 100],
    linestyle="--",
    linewidth=2
)

# ------------------------------------------------------------
# Traffic Lights
# ------------------------------------------------------------

# North
ax.add_patch(
    patches.Rectangle(
        (68, 68),
        8,
        20,
        facecolor="black"
    )
)

# South
ax.add_patch(
    patches.Rectangle(
        (24, 12),
        8,
        20,
        facecolor="black"
    )
)

# East
ax.add_patch(
    patches.Rectangle(
        (68, 24),
        20,
        8,
        facecolor="black"
    )
)

# West
ax.add_patch(
    patches.Rectangle(
        (12, 68),
        20,
        8,
        facecolor="black"
    )
)

# ------------------------------------------------------------
# Current Traffic State
# ------------------------------------------------------------

traffic_state = random.choice([0, 1, 2])

if traffic_state == 0:
    traffic_text = "LOW TRAFFIC"

elif traffic_state == 1:
    traffic_text = "MEDIUM TRAFFIC"

else:
    traffic_text = "HIGH TRAFFIC"


# ------------------------------------------------------------
# Q-Learning Best Action
# ------------------------------------------------------------

best_action = np.argmax(q_table[traffic_state])

if best_action == 0:
    action_text = "KEEP SIGNAL"

else:
    action_text = "CHANGE SIGNAL"


# ------------------------------------------------------------
# Vehicles
# ------------------------------------------------------------

# Horizontal vehicles
for i in range(8):

    x = random.randint(5, 30)
    y = random.choice([42, 55])

    vehicle = patches.Rectangle(
        (x, y),
        6,
        3,
        facecolor="red"
    )

    ax.add_patch(vehicle)


# Vertical vehicles
for i in range(8):

    x = random.choice([42, 55])
    y = random.randint(5, 30)

    vehicle = patches.Rectangle(
        (x, y),
        3,
        6,
        facecolor="blue"
    )

    ax.add_patch(vehicle)


# ------------------------------------------------------------
# Traffic Signal Colors
# ------------------------------------------------------------

if best_action == 0:

    signal_color = "green"

else:

    signal_color = "red"


# Main traffic signal
ax.add_patch(
    patches.Circle(
        (50, 72),
        4,
        facecolor=signal_color,
        edgecolor="black",
        linewidth=2
    )
)


# ------------------------------------------------------------
# Information Panel
# ------------------------------------------------------------

ax.text(
    50,
    96,
    "INTELLIGENT TRAFFIC SIGNAL OPTIMIZATION",
    ha="center",
    fontsize=18,
    fontweight="bold"
)

ax.text(
    50,
    91,
    "Q-Learning Traffic Control System",
    ha="center",
    fontsize=13
)

ax.text(
    5,
    5,
    f"Traffic Level : {traffic_text}\n"
    f"Best Action   : {action_text}\n"
    f"Q-Value       : {q_table[traffic_state, best_action]:.2f}",
    fontsize=12,
    bbox=dict(
        boxstyle="round",
        facecolor="white",
        edgecolor="black"
    )
)

# ------------------------------------------------------------
# Labels
# ------------------------------------------------------------

ax.text(
    50,
    82,
    "TRAFFIC SIGNAL",
    ha="center",
    fontsize=10,
    fontweight="bold"
)

ax.text(
    8,
    50,
    "WEST",
    fontsize=11,
    fontweight="bold"
)

ax.text(
    87,
    50,
    "EAST",
    fontsize=11,
    fontweight="bold"
)

ax.text(
    50,
    8,
    "SOUTH",
    fontsize=11,
    fontweight="bold"
)

ax.text(
    50,
    90,
    "NORTH",
    fontsize=11,
    fontweight="bold"
)


# ------------------------------------------------------------
# Remove Axis
# ------------------------------------------------------------

ax.set_xticks([])
ax.set_yticks([])

for spine in ax.spines.values():
    spine.set_visible(False)

plt.title(
    "Real-Time Traffic Intersection Visualization",
    fontsize=14
)

plt.show()
```
